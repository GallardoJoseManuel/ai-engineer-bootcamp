```python
import numpy as np

a = np.array([10,20,30,40,50])
print(a.shape)      # Pair (rows x columns)
print(a.ndim)       # Number of dimensions of the array
print(a.size)       # Total number of elements
print(a.dtype)      # Type of the elements contained
print(a.itemsize)   # Size of a singular item measured in bytes
print(a.nbytes)     # total size of memory used (8x5)
```
