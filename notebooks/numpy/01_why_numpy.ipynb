# Why NumPy?

## Objective

Understand why NumPy exists and why it is the foundation of modern Artificial Intelligence.

```python
numbers = [1, 2, 3, 4, 5]

print(numbers * 2)
```

```python
numbers + numbers
```

```python
for i in range(len(numbers)):
    numbers[i] *= 2

print(numbers)
```

```python
import numpy as np

numbers = np.array([1, 2, 3, 4, 5])

print(numbers)
```

```python
numbers * 2
```

```python
numbers + numbers
```

```python
numbers ** 2
```

# Key Takeaways

- Python lists store references to Python objects rather than raw values.
- NumPy arrays store homogeneous data in contiguous memory, enabling highly efficient numerical computations.
- Vectorized operations eliminate the need for explicit Python loops, resulting in much better performance.
- NumPy is the foundation of most scientific computing and modern AI libraries such as pandas, PyTorch and TensorFlow.
