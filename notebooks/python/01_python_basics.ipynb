# Python Basics


```python
x = 10
print(x)
```

## Variables


'x' and 'y' are pointers to the object that contains the value 10



```python
x = 10
y = x

print(x)
print(y)
```

## References and Objects


'y' still points to the old value 10 while 'x' points to a new object containing the new value 20

```python
x = 10
y = x

x = 20

print(x)
print(y)
```

## Mutable vs Immutable Objects



```python
a = [1, 2, 3]

b = a

b.append(4)

print(a)
print(b)
```

```python
a = [1,2,3]

b = a.copy()

b.append(4)

print(a)
print(b)
```

copy() -> shallow Copy    
In this case we create a new list which is referenced by b. The internal copy does not copy the content of the 2 internal list but the references of these. 
This results in handling the same references of the internal lists when operating with a or b

```python
a = [[1, 2], [3, 4]]

b = a.copy()

b[0].append(99)

print(a)
print(b)
```

# Lesson learned

In Python, variables do not store values directly.

They store references to objects.

Assigning one variable to another copies the reference, not necessarily the object itself.
